#### Based on the dataset description provided on Kaggle, the meaning of each column in the dataset is as follows:

- customer_id: Unique identifier assigned to each customer.
- gender: Gender of the customer (e.g., Male, Female, Other).
- age: Age of the customer in years.
- country: Country where the customer is located.
- city: City where the customer resides.
- customer_segment: Business-defined customer classification (e.g., SMB, Enterprise, Individual).
- tenure_months: Number of months the customer has been subscribed to the service.
- signup_channel: Channel through which the customer signed up (e.g., Organic, Ads, Referral).
- contract_type: Type of subscription contract (e.g., Monthly, Quarterly, Annual).

Product Usage Metrics
- monthly_logins: Total number of times the customer logged into the platform in a month.
- weekly_active_days: Average number of days per week the customer actively used the platform.
- avg_session_time: Average duration of each user session (typically in minutes).
- features_used: Number of distinct product features used by the customer.
- usage_growth_rate: Percentage change in usage over a defined recent period.
- last_login_days_ago: Number of days since the customer last logged in.

Billing & Payment Information
- monthly_fee: Subscription fee charged to the customer per month.
- total_revenue: Total revenue generated from the customer over their lifetime.
- payment_method: Method used for payment (e.g., Credit Card, PayPal, Bank Transfer).
- payment_failures: Number of failed payment attempts recorded.
- discount_applied: Indicates whether a discount has been applied to the customer’s subscription (Yes/No or binary flag).
- price_increase_last_3m: Indicates whether the customer experienced a price increase in the last three months (Yes/No or binary flag).

Customer Support & Satisfaction
- support_tickets: Number of customer support tickets raised.
- avg_resolution_time: Average time taken to resolve support tickets (e.g., in hours or days).
- complaint_type: Category of complaint raised by the customer (e.g., Billing, Technical, Service).
- csat_score: Customer Satisfaction Score (typically on a defined scale such as 1–5).
- escalations: Number of support cases escalated beyond first-level support.

Engagement & Feedback
- email_open_rate: Percentage of marketing or communication emails opened by the customer.
- marketing_click_rate: Percentage of marketing emails where links were clicked.
- nps_score: Net Promoter Score provided by the customer (typically ranging from -100 to 100 or 0 to 10 depending on scale).
- survey_response: Indicates whether the customer responded to feedback surveys (Yes/No or binary flag).
- referral_count: Number of new customers referred by the existing customer.

Critical Variable
- churn: Binary target variable indicating whether the customer discontinued the service (0 = No, 1 = Yes).

#### Observations on Data Types

- Clearly, most columns in the dataset are stored in appropriate data types for analysis. However, a few adjustments would improve analytical clarity, memory efficiency, and modeling readiness.

- For example, churn is stored as int64, but it represents a binary categorical outcome (0 = No, 1 = Yes). While keeping it as integer is acceptable for machine learning models, converting it to a categorical type is more appropriate for exploratory data analysis and visualization.

- Similarly, nps_score is stored as int64. Although numeric, it often behaves like an ordinal or categorical metric in business analysis (e.g., Detractors, Passives, Promoters). Depending on the analysis goal, it may require grouping or transformation.

- Additionally, several categorical variables such as gender, country, city, customer_segment, signup_channel, contract_type, payment_method, complaint_type, discount_applied, price_increase_last_3m, and survey_response are stored as object (string) types. While technically correct, converting them to the category data type would improve memory usage and analytical performance.

#### Checking for any missing values per column

```python
# Count of NaN / None values per column
nan_count = df.isnull().sum()

# Count of empty strings per column
empty_count = (df == '').sum()

# Total missing values (NaN + empty strings)
total_missing = nan_count + empty_count

# Percentage of missing values
missing_percent = (total_missing / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Values': total_missing,
    'Percentage (%)': missing_percent
}).sort_values(by='Percentage (%)', ascending=False)

print(missing_df)
```

#### Summary from the Observations

Columns with Missing Values
- complaint_type - 2,045 missing values (20.45%) 


For 'complaint_type', replacing NaN with the mode is NOT recommended, because NaN here has a business meaning: it indicates the customer did NOT make any complaint.

Imputing with the most common complaint type would distort the data and bias analysis.

Instead, we replace NaN with a meaningful category, e.g., "No Complaint", which preserves the business interpretation and allows easy analysis and modeling.


```python
df['complaint_type'] = df['complaint_type'].fillna('No Complaint')
```
